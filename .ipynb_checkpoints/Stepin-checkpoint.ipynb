```python
from sklearn.linear_model import LinearRegression
import numpy as np

# House sizes in square feet
sizes = np.array([500, 750, 1000, 1250, 1500]).reshape(-1, 1)

#House prices in Kshs.
prices = np.array([15000, 20000, 25000, 30000, 35000])

#create and train the model
model = LinearRegression()
model.fit(sizes, prices)

#predict price for a 1100 sq ft house
prediction = model.predict([[1100]])
print(f"Predicted price: Kshs.{prediction[0]:.1f}k")
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Generate a smooth line for the model
line_sizes = np.linspace(400, 1600, 100).reshape(-1, 1)
line_prices = model.predict(line_sizes)

# Plot original data points
plt.scatter(sizes, prices, color='blue', label='Actual data')

# Plot the model's line
plt.plot(line_sizes, line_prices, color='red', label='Model prediction line')

# Plot our prediction
plt.scatter(1100, prediction[0], color='green', s=100, label='Our prediction')

plt.title("House Prices - Linear Regression")
plt.xlabel("Size (sqft)")
plt.ylabel("Price (Kshs.000s)")
plt.legend()
plt.show()
```

```python
from sklearn.metrics import r2_score

predictions = model.predict(sizes)
score = r2_score(prices, predictions)
print(f"Model accuracy: {score:.2f}")
```

```python
from sklearn.linear_model import LogisticRegression
import numpy as np

# Hours studied
hours = np.array([1, 2, 3, 4, 5, 6, 7, 8]).reshape(-1, 1)

# Pass or fail (0 = fail, 1 = pass)
results = np.array([0, 0, 0, 0, 1, 1, 1, 1])

# Create and train the model
model = LogisticRegression()
model.fit(hours, results)

# Predict if a student who studied 5.5 hours will pass
prediction = model.predict([[5.5]])
print(f"Prediction: {'Pass' if prediction[0] == 1 else 'Fail'}")
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Generate smooth line
x = np.linspace(0, 9, 300).reshape(-1, 1)
y = model.predict_proba(x)[:, 1]

# Plot the curve
plt.plot(x, y, color='red', label='Probability of passing')

# Plot original data points
plt.scatter(hours, results, color='blue', label='Actual data', zorder=5)

# Plot our prediction
plt.scatter(5.5, model.predict_proba([[5.5]])[0][1], color='green', s=100, label='Our prediction', zorder=5)

plt.title("Logistic Regression - Pass/Fail Prediction")
plt.xlabel("Hours Studied")
plt.ylabel("Probability of Passing")
plt.legend()
plt.show()
```

```python
from sklearn.tree import DecisionTreeClassifier

# Hours studied and hours slept
X = np.array([
    [2, 5], [3, 4], [4, 6], [5, 7],
    [6, 5], [7, 8], [8, 6], [1, 3]
])

# Pass or fail
y = np.array([0, 0, 0, 1, 1, 1, 1, 0])

# Create and train the model
model = DecisionTreeClassifier()
model.fit(X, y)

# Predict for a student who studied 5 hours and slept 7 hours
prediction = model.predict([[5, 7]])
print(f"Prediction: {'Pass' if prediction[0] == 1 else 'Fail'}")
```

```python
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plot_tree(model, 
          feature_names=['Hours Studied', 'Hours Slept'],
          class_names=['Fail', 'Pass'],
          filled=True)
plt.title("Decision Tree")
plt.show()
```

```python
from sklearn.neighbors import KNeighborsClassifier

# Same data as before - hours studied and hours slept
X = np.array([
    [2, 5], [3, 4], [4, 6], [5, 7],
    [6, 5], [7, 8], [8, 6], [1, 3]
])

y = np.array([0, 0, 0, 1, 1, 1, 1, 0])

# Create model - k=3 means look at 3 nearest neighbors
model = KNeighborsClassifier(n_neighbors=3)
model.fit(X, y)

# Predict for a student who studied 5 hours and slept 7 hours
prediction = model.predict([[5, 7]])
print(f"Prediction: {'Pass' if prediction[0] == 1 else 'Fail'}")
```

```python
import pandas as pd

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)
print(df.shape)
```

```python
print(df.head())
```

```python
print(df.isnull().sum())
```

```python
# Fill missing ages with the average age
df['Age'].fillna(df['Age'].mean(), inplace=True)

# Drop the Cabin column entirely
df.drop(columns=['Cabin'], inplace=True)

# Fill missing Embarked with most common port
df['Embarked'].fillna(df['Embarked'].mode()[0], inplace=True)

# Verify no more missing values
print(df.isnull().sum())
```
